```python
import numpy as np

# Create two vectors
A = np.array([2, 4, 6])
B = np.array([1, 3, 5])

# Vector Addition
addition = A + B

# Vector Subtraction
subtraction = A - B

# Scalar Multiplication
scalar = 3
scalar_multiplication = scalar * A

# Dot Product
dot_product = np.dot(A, B)

# Vector Magnitude
magnitude_A = np.linalg.norm(A)
magnitude_B = np.linalg.norm(B)

# Display results
print("Vector A:", A)
print("Vector B:", B)
print("Vector Addition:", addition)
print("Vector Subtraction:", subtraction)
print("Scalar Multiplication:", scalar_multiplication)
print("Dot Product:", dot_product)
print("Magnitude of A:", magnitude_A)
print("Magnitude of B:", magnitude_B)
```

```python
import numpy as np

# Create a matrix
A = np.array([[1, 2],
              [3, 4]])

# Calculate inverse
inverse_A = np.linalg.inv(A)

# Display results
print("Matrix A:")
print(A)

print("\nInverse of Matrix A:")
print(inverse_A)

# Verify A × A^-1 = Identity Matrix
print("\nA × A^-1:")
print(np.dot(A, inverse_A))
```

```python
import numpy as np

# Create a matrix
A = np.array([[1, 2],
              [3, 4]])

# Determinant
determinant = np.linalg.det(A)

# Rank
rank = np.linalg.matrix_rank(A)

# Trace
trace = np.trace(A)

# Display results
print("Matrix A:")
print(A)

print("\nDeterminant:", determinant)
print("Rank:", rank)
print("Trace:", trace)
```

```python
import numpy as np

# Create a square matrix
A = np.array([[2, 1],
              [1, 2]])

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

# Display results
print("Matrix A:")
print(A)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```
