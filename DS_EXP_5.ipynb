```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load the dataset
df = pd.read_csv("employee_performance.csv")

# Select numerical features
features = ["Age", "Experience", "Salary"]

# Display original dataset
print("Original Dataset:")
display(df[features])

# Create StandardScaler
scaler = StandardScaler()

# Perform Z-score Normalization
df_normalized = df.copy()
df_normalized[features] = scaler.fit_transform(df[features])

# Display normalized dataset
print("Dataset After Z-score Normalization:")
display(df_normalized[features])

# Compare original and normalized values
comparison = pd.DataFrame()

for feature in features:
    comparison[feature + "_Original"] = df[feature]
    comparison[feature + "_Normalized"] = df_normalized[feature]

print("Comparison of Original and Normalized Values:")
display(comparison)
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("employee_performance.csv")

# Display original Salary values
print("Original Salary Values:")
display(df[["Salary"]])

# Perform equal-width binning
# Divide Salary into 4 equal-width intervals
df["Salary_Bin"] = pd.cut(
    df["Salary"],
    bins=4,
    labels=["Low", "Medium", "High", "Very High"]
)

# Display the result
print("Salary After Equal-Width Binning:")
display(df[["Salary", "Salary_Bin"]])

# Display the number of employees in each bin
print("Number of Employees in Each Salary Bin:")
print(df["Salary_Bin"].value_counts().sort_index())
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("employee_performance.csv")

# Display original Experience values
print("Original Experience Values:")
display(df[["Experience"]])

# Perform equal-frequency binning
# Divide Experience into 4 groups with approximately equal number of records
df["Experience_Bin"] = pd.qcut(
    df["Experience"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)

# Display the generated bins
print("Experience After Equal-Frequency Binning:")
display(df[["Experience", "Experience_Bin"]])

# Display the number of records in each bin
print("Number of Employees in Each Bin:")
print(df["Experience_Bin"].value_counts().sort_index())
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load the dataset
df = pd.read_csv("employee_performance.csv")

# Create Performance category from Performance_Score
df["Performance"] = pd.cut(
    df["Performance_Score"],
    bins=[0, 79, 89, 100],
    labels=["Low", "Medium", "High"]
)

# Create LabelEncoder
encoder = LabelEncoder()

# Encode Department
df["Department_Encoded"] = encoder.fit_transform(df["Department"])

# Encode Performance
df["Performance_Encoded"] = encoder.fit_transform(df["Performance"])

# Display original and encoded values
print("Label Encoded Dataset:")
display(
    df[
        [
            "Department",
            "Department_Encoded",
            "Performance",
            "Performance_Encoded"
        ]
    ]
)
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Load the dataset
df = pd.read_csv("employee_performance.csv")

# Create Performance category from Performance_Score
df["Performance"] = pd.cut(
    df["Performance_Score"],
    bins=[0, 79, 89, 100],
    labels=["Low", "Medium", "High"]
)

# Create Ordinal Encoder
encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

# Perform Ordinal Encoding
df["Performance_Encoded"] = encoder.fit_transform(
    df[["Performance"]]
).astype(int)

# Display original and encoded values
print("Performance Before and After Ordinal Encoding:")
display(
    df[
        [
            "Performance",
            "Performance_Encoded"
        ]
    ]
)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load the dataset
df = pd.read_csv("employee_performance.csv")

# Select numerical attributes
features = [
    "Age",
    "Experience",
    "Salary",
    "Performance_Score"
]

X = df[features]

# Standardize the numerical data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Perform PCA and reduce to 2 principal components
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

# Create a DataFrame for the transformed feature matrix
pca_df = pd.DataFrame(
    X_pca,
    columns=["Principal_Component_1", "Principal_Component_2"]
)

# Display the transformed feature matrix
print("Transformed Feature Matrix after PCA:")
display(pca_df)

# Display explained variance
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Display total explained variance
print("\nTotal Explained Variance:")
print(pca.explained_variance_ratio_.sum())
```
