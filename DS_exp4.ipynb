```python
import pandas as pd
from scipy.spatial.distance import euclidean

# Create Student Performance Dataset
data = {
    'Student': ['A', 'B', 'C', 'D', 'E'],
    'Math': [85, 78, 92, 70, 88],
    'Science': [90, 75, 95, 68, 85],
    'English': [80, 72, 91, 65, 87]
}

# Convert to DataFrame
df = pd.DataFrame(data)

print("Student Performance Dataset")
print(df)

# Select Student A and Student B
student1 = df[df['Student'] == 'A'][
    ['Math', 'Science', 'English']
].values[0]

student2 = df[df['Student'] == 'B'][
    ['Math', 'Science', 'English']
].values[0]

# Calculate Euclidean Distance
distance = euclidean(student1, student2)

print("\nEuclidean Distance between Student A and Student B:")
print(round(distance, 2))

# Calculate distance between Student A and all students
print("\nSimilarity of Student A with other students")

for index, row in df.iterrows():
    current_student = row[['Math', 'Science', 'English']].values
    dist = euclidean(student1, current_student)

    print(f"Distance between A and {row['Student']} = {dist:.2f}")
```

```python
import pandas as pd
from scipy.spatial.distance import cityblock

# Create Healthcare Dataset
data = {
    'Patient': ['P1', 'P2', 'P3', 'P4', 'P5'],
    'Age': [25, 40, 35, 50, 45],
    'BloodPressure': [120, 140, 130, 150, 145],
    'Cholesterol': [180, 220, 200, 240, 230]
}

# Convert to DataFrame
df = pd.DataFrame(data)

print("Healthcare Dataset")
print(df)

# Select Patient P1 and P2
patient1 = df[df['Patient'] == 'P1'][
    ['Age', 'BloodPressure', 'Cholesterol']
].values[0]

patient2 = df[df['Patient'] == 'P2'][
    ['Age', 'BloodPressure', 'Cholesterol']
].values[0]

# Calculate Manhattan Distance
distance = cityblock(patient1, patient2)

print("\nManhattan Distance between P1 and P2:")
print(distance)

# Calculate distance between P1 and all patients
print("\nDissimilarity of Patient P1 with other patients")

for index, row in df.iterrows():
    patient = row[['Age', 'BloodPressure', 'Cholesterol']].values
    dist = cityblock(patient1, patient)

    print(f"Distance between P1 and {row['Patient']} = {dist}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Create Text Dataset
documents = [
    "Artificial Intelligence is transforming healthcare.",
    "Artificial Intelligence improves healthcare systems.",
    "Machine Learning is a branch of Artificial Intelligence.",
    "Football is a popular outdoor sport.",
    "Healthcare uses AI for disease prediction."
]

# Create DataFrame
df = pd.DataFrame({
    "Document": documents
})

print("Text Dataset")
print(df)

# Convert text into numerical vectors
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)

# Compute Cosine Similarity Matrix
similarity_matrix = cosine_similarity(X)

print("\nCosine Similarity Matrix")

print(pd.DataFrame(
    similarity_matrix,
    columns=[f'Doc{i+1}' for i in range(len(documents))],
    index=[f'Doc{i+1}' for i in range(len(documents))]
))

# Similarity between Document 1 and Document 2
print("\nSimilarity between Document 1 and Document 2:")
print(round(similarity_matrix[0][1], 2))

# Similarity of Document 1 with all documents
print("\nSimilarity of Document 1 with all Documents")

for i in range(len(documents)):
    print(f"Document 1 and Document {i+1}: {similarity_matrix[0][i]:.2f}")
```

```python
import pandas as pd

# Create Transaction Dataset
data = {
    "Transaction": ["T1", "T2", "T3", "T4"],
    "Milk": [1, 1, 0, 1],
    "Bread": [1, 0, 1, 1],
    "Butter": [0, 1, 1, 1],
    "Eggs": [1, 0, 1, 0],
    "Cheese": [0, 1, 0, 1]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Categorical Transaction Dataset")
print(df)

# Select T1 and T2
t1 = df.iloc[0, 1:].astype(int).values
t2 = df.iloc[1, 1:].astype(int).values

# Function to calculate Jaccard Similarity
def jaccard_similarity(a, b):
    intersection = sum((a == 1) & (b == 1))
    union = sum((a == 1) | (b == 1))

    if union == 0:
        return 0

    return intersection / union

# Similarity between T1 and T2
similarity = jaccard_similarity(t1, t2)

print("\nJaccard Similarity between T1 and T2:")
print(round(similarity, 2))

# Similarity between T1 and all transactions
print("\nSimilarity of T1 with other transactions")

for index, row in df.iterrows():
    transaction = row.iloc[1:].astype(int).values
    sim = jaccard_similarity(t1, transaction)

    print(
        "Similarity between T1 and",
        row["Transaction"],
        "=",
        round(sim, 2)
    )
```
