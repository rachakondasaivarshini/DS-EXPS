```python
import numpy as np
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])
addition = a + b
subtraction = b - a
multiplication = a * b
division = b / a
print("Array a:", a)
print("Array b:", b)
print("\nAddition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```
