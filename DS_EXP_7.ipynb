```python
import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO

# Create the sales dataset
data = """Sales,Profit,Quantity
1000,200,10
1500,300,15
2000,450,20
2500,500,25
3000,700,30
3500,800,35
4000,950,40
4500,1000,45
5000,1200,50
5500,1300,55
"""

# Load the dataset
df = pd.read_csv(StringIO(data))

# Display first five records
print(df.head())

# Create scatter plots
plt.figure(figsize=(15, 4))

# 1. Sales vs Profit
plt.subplot(1, 3, 1)
plt.scatter(df["Sales"], df["Profit"])
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")
plt.grid(True)

# 2. Sales vs Quantity
plt.subplot(1, 3, 2)
plt.scatter(df["Sales"], df["Quantity"])
plt.xlabel("Sales")
plt.ylabel("Quantity")
plt.title("Sales vs Quantity")
plt.grid(True)

# 3. Profit vs Quantity
plt.subplot(1, 3, 3)
plt.scatter(df["Profit"], df["Quantity"])
plt.xlabel("Profit")
plt.ylabel("Quantity")
plt.title("Profit vs Quantity")
plt.grid(True)

# Adjust layout
plt.tight_layout()

# Display plots
plt.show()
```
