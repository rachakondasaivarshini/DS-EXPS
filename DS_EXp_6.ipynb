```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the sales dataset
df = pd.read_csv("sales_data.csv")

# Display Sales and Profit values
print("Sales and Profit Values:")
display(df[["Sales", "Profit"]])

# Create Scatter Plot
plt.figure(figsize=(8, 5))

plt.scatter(df["Sales"], df["Profit"])

plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Scatter Plot: Sales vs Profit")

plt.grid(True)

# Display the scatter plot
plt.show()

# Calculate correlation
correlation = df["Sales"].corr(df["Profit"])

print("\nCorrelation between Sales and Profit:")
print(correlation)

# Analyze the relationship
if correlation > 0:
    print("There is a positive relationship between Sales and Profit.")
elif correlation < 0:
    print("There is a negative relationship between Sales and Profit.")
else:
    print("There is no linear relationship between Sales and Profit.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create sample sales data
data = {
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone",
                "Tablet", "Laptop", "Phone", "Tablet", "Laptop"],

    "Sales": [50000, 30000, 20000, 55000, 32000,
              22000, 60000, 35000, 25000, 65000],

    "Region": ["North", "South", "East", "West", "North",
               "South", "East", "West", "North", "South"]
}

df = pd.DataFrame(data)

# Calculate total sales for each region
region_sales = df.groupby("Region")["Sales"].sum()

print("Total Sales Across Regions:")
print(region_sales)

# Create Bar Chart
plt.figure(figsize=(8, 5))

plt.bar(region_sales.index, region_sales.values)

plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.title("Total Sales Across Different Regions")

plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Create sample sales data
data = {
    "Product": [
        "Laptop", "Phone", "Tablet", "Laptop", "Phone",
        "Tablet", "Laptop", "Phone", "Tablet", "Laptop"
    ],

    "Sales": [
        50000, 30000, 20000, 55000, 32000,
        22000, 60000, 35000, 25000, 65000
    ],

    "Profit": [
        8000, 5000, 3500, 9000, 5500,
        4000, 10000, 6000, 4500, 11000
    ],

    "Region": [
        "North", "South", "East", "West", "North",
        "South", "East", "West", "North", "South"
    ]
}

df = pd.DataFrame(data)

# Display original data
print("Original Sales Data:")
display(df)

# Create Pivot Table
pivot_table = pd.pivot_table(
    df,
    values=["Sales", "Profit"],
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

# Display Pivot Table
print("Pivot Table - Total Sales and Profit by Region and Product:")
display(pivot_table)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample sales data
data = {
    "Product": [
        "Laptop", "Phone", "Tablet", "Laptop", "Phone",
        "Tablet", "Laptop", "Phone", "Tablet", "Laptop"
    ],

    "Sales": [
        50000, 30000, 20000, 55000, 32000,
        22000, 60000, 35000, 25000, 65000
    ],

    "Quantity": [
        5, 10, 8, 6, 12,
        9, 7, 15, 10, 8
    ],

    "Profit": [
        8000, 5000, 3500, 9000, 5500,
        4000, 10000, 6000, 4500, 11000
    ],

    "Region": [
        "North", "South", "East", "West", "North",
        "South", "East", "West", "North", "South"
    ]
}

df = pd.DataFrame(data)

# Select numerical attributes
numerical_data = df[["Sales", "Profit", "Quantity"]]

# Calculate correlation matrix
correlation_matrix = numerical_data.corr()

# Display correlation matrix
print("Correlation Matrix:")
display(correlation_matrix)

# Create Heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Sales, Profit and Quantity")
plt.tight_layout()
plt.show()
```
