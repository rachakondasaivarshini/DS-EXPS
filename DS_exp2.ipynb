```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [22, 25, 23, 24, 21],
    "Marks": [85, 90, 78, 88, 95]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

print("\nFirst 3 Records:")
print(df.head(3))

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nStudents Scoring Above 85:")
print(df[df["Marks"] > 85])

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])

print("Array A:")
print(a)

print("Array B:")
print(b)

print("Addition:")
print(a + b)

print("Multiplication:")
print(a * b)

print("Mean:")
print(np.mean(a))

print("Maximum:")
print(np.max(a))

print("Standard Deviation:")
print(np.std(a))

matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Matrix:")
print(matrix)

print("Transpose:")
print(matrix.T)
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 170, 160, 200]

plt.figure(figsize=(8, 5))
plt.plot(months, sales, marker='o', linewidth=2)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Hours": [2, 3, 4, 5, 6, 7, 8, 9],
    "Marks": [35, 40, 50, 60, 65, 75, 85, 95]
}

df = pd.DataFrame(data)

sns.set_style("whitegrid")

plt.figure(figsize=(8, 5))
sns.regplot(data=df, x="Hours", y="Marks")

plt.title("Study Hours vs Marks")
plt.show()
```

```python
import plotly.express as px
import pandas as pd

data = {
    "Department": ["HR", "Sales", "IT", "Finance"],
    "Employees": [15, 25, 35, 20]
}

df = pd.DataFrame(data)

fig = px.bar(
    df,
    x="Department",
    y="Employees",
    color="Department",
    title="Employees in Each Department"
)

fig.show()
```

```python
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

output_notebook()

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 170, 160, 200]

p = figure(
    title="Monthly Sales",
    x_range=months,
    width=700,
    height=400
)

p.line(months, sales, line_width=3)
p.circle(months, sales, size=10)

show(p)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[6]])

print("Prediction for X = 6:", prediction[0])
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
import pandas as pd
import statsmodels.api as sm

data = {
    "Hours": [2, 3, 4, 5, 6, 7, 8],
    "Marks": [30, 40, 50, 60, 70, 80, 90]
}

df = pd.DataFrame(data)

X = sm.add_constant(df["Hours"])
Y = df["Marks"]

model = sm.OLS(Y, X).fit()

print(model.summary())
```
