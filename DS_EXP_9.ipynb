```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli

p = 0.7
x = np.array([0, 1])

pmf = bernoulli.pmf(x, p)

print("Bernoulli Distribution")
print("Success:", p)
print("Failure:", 1-p)

for value, probability in zip(x, pmf):
    print(f"P(X={value}) = {probability:.2f}")

samples = np.random.binomial(1, p, 10)
print("Random Samples:", samples)

plt.bar(x, pmf)
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.5

x = np.arange(0, n+1)
probabilities = binom.pmf(x, n, p)

print("Binomial Distribution")

for outcome, probability in zip(x, probabilities):
    print(f"Successes = {outcome}: {probability:.4f}")

plt.bar(x, probabilities)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lam = 4

x = np.arange(0, 15)
probabilities = poisson.pmf(x, lam)

print("Poisson Distribution")

for value, probability in zip(x, probabilities):
    print(f"P(X={value}) = {probability:.4f}")

plt.bar(x, probabilities)
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 70
std = 10

x = np.linspace(40, 100, 100)
y = norm.pdf(x, mean, std)

plt.plot(x, y)
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")
plt.show()
```
